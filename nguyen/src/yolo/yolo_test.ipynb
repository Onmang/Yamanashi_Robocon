# 自動アノテーション
事前に学習した重みbest.ptで新しいデータのラベリング

```python
import torch
print(torch.cuda.is_available()) 
# True と出れば、ローカルのGPUを使って爆速で学習できます！
```

```python
from ultralytics import YOLO
import os
```

```python
# 1. モデルのパス（ローカルPC上のパスに書き換え）
# 例: 'C:/Users/Name/Downloads/best.pt' など
model_path = 'best.pt' 
model = YOLO(model_path)

# 2. アノテーションしたい画像が入っているフォルダ（ローカルのパス）
# 同じフォルダに 'new_images' フォルダを作っておくと楽です
source_dir = "C:\\Users\\guenk\\yolo\\datasets\\ichgan0622"
```

**model.predict** の引数

推論コントロール（基本設定）
source: 入力ソース（画像フォルダ、動画パス、USBカメラのID 0、URLなど）

conf: 信頼度のしきい値（デフォルト 0.25）。これ未満の検出は無視

iou: NMS（重複枠の除去）のしきい値。値が小さいほど重複に厳しくなる

imgsz: 推論時の画像サイズ（例: 640）。学習時と同じサイズが推奨

half: FP16（半精度）を使用。高速化とメモリ節約に有効

```python
# 3. 自動アノテーション（推論）を実行
results = model.predict(
    source=source_dir, 
    conf=0.25,      
    save_txt=True,  # 予測結果をtxtで保存
    save_conf=False,
    project='auto_ann', # ローカルの作業フォルダ内に 'auto_ann' が作られます
    name='predict_labels',
    device=0        # 明示的にGPU（CUDA）を使う指定
)

print("自動アノテーションが完了しました。'auto_ann/predict_labels/labels' を確認してください。")
```
